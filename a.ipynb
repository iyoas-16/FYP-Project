```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report, roc_auc_score, roc_curve
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras.utils import to_categorical

# --- Load dataset ---
df = pd.read_csv('phishing_url.csv')

# --- Basic checks (optional) ---
# print(df.columns)
# Ensure 'url', 'target', 'Phishing' exist in your dataset
assert 'url' in df.columns, "'url' column missing"
assert 'target' in df.columns, "'target' column missing"
assert 'Phishing' in df.columns, "'Phishing' column missing"

# --- Prepare features ---
# Combine 'url' and 'target' (as you did)
df['combined'] = df['url'].astype(str) + ' ' + df['target'].astype(str)

# Vectorize (CountVectorizer)
vectorizer = CountVectorizer(stop_words='english', max_features=20000)  # limit features for memory control
X_counts = vectorizer.fit_transform(df['combined'])
y = df['Phishing'].astype(int).values  # ensure integer labels (0/1)

# Save vectorizer for later use
joblib.dump(vectorizer, 'vectorizer.pkl')

# --- Train/test split ---
X_train, X_test, y_train, y_test = train_test_split(X_counts, y, test_size=0.2, random_state=42, stratify=y)

# Convert sparse matrices to dense arrays for Keras (DecisionTree can work with sparse but we convert once)
# WARNING: if you have very high dimensional data and very large dataset, converting to dense may use a lot of memory.
X_train_arr = X_train.toarray()
X_test_arr = X_test.toarray()

# --- 1) Decision Tree (your original classifier) ---
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train_arr, y_train)
joblib.dump(clf, 'decision_tree_model.pkl')

y_pred_dt = clf.predict(X_test_arr)
acc_dt = accuracy_score(y_test, y_pred_dt)
roc_dt = roc_auc_score(y_test, clf.predict_proba(X_test_arr)[:, 1]) if hasattr(clf, "predict_proba") else np.nan
print(f"Decision Tree Accuracy: {acc_dt:.4f}, ROC AUC: {roc_dt:.4f}")

# --- 2) Deep Learning Model (Dense Neural Network) ---
input_dim = X_train_arr.shape[1]

# Build a simple feed-forward neural network
def build_deep_model(input_dim):
    model = Sequential()
    model.add(Dense(512, activation='relu', input_shape=(input_dim,)))
    model.add(Dropout(0.4))
    model.add(Dense(256, activation='relu'))
    model.add(Dropout(0.3))
    model.add(Dense(64, activation='relu'))
    model.add(Dropout(0.2))
    model.add(Dense(1, activation='sigmoid'))  # binary classification
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

deep_model = build_deep_model(input_dim)

# Early stopping to avoid overfitting
es = EarlyStopping(monitor='val_loss', patience=3, restore_best_weights=True, verbose=1)

history = deep_model.fit(
    X_train_arr, y_train,
    validation_split=0.1,
    epochs=20,
    batch_size=64,
    callbacks=[es],
    verbose=2
)

# Save deep model
deep_model.save('deep_model.h5')

# Predictions & evaluation
y_pred_dl_prob = deep_model.predict(X_test_arr).ravel()
y_pred_dl = (y_pred_dl_prob >= 0.5).astype(int)

acc_dl = accuracy_score(y_test, y_pred_dl)
roc_dl = roc_auc_score(y_test, y_pred_dl_prob)
print(f"Deep NN Accuracy: {acc_dl:.4f}, ROC AUC: {roc_dl:.4f}")

# --- Detailed reports ---
print("\nDecision Tree Classification Report:")
print(classification_report(y_test, y_pred_dt))
print("\nDeep NN Classification Report:")
print(classification_report(y_test, y_pred_dl))

# --- Confusion matrices (side-by-side) ---
cm_dt = confusion_matrix(y_test, y_pred_dt)
cm_dl = confusion_matrix(y_test, y_pred_dl)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
sns.heatmap(cm_dt, annot=True, fmt='d', ax=axes[0])
axes[0].set_title('Decision Tree Confusion Matrix')
axes[0].set_xlabel('Predicted')
axes[0].set_ylabel('True')

sns.heatmap(cm_dl, annot=True, fmt='d', ax=axes[1])
axes[1].set_title('Deep NN Confusion Matrix')
axes[1].set_xlabel('Predicted')
axes[1].set_ylabel('True')

plt.tight_layout()
plt.show()

# --- ROC curves ---
fpr_dt, tpr_dt, _ = roc_curve(y_test, clf.predict_proba(X_test_arr)[:, 1])
fpr_dl, tpr_dl, _ = roc_curve(y_test, y_pred_dl_prob)

plt.figure(figsize=(7, 6))
plt.plot(fpr_dt, tpr_dt, label=f'Decision Tree (AUC = {roc_dt:.3f})')
plt.plot(fpr_dl, tpr_dl, label=f'Deep NN (AUC = {roc_dl:.3f})')
plt.plot([0, 1], [0, 1], '--', label='Random')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve Comparison')
plt.legend()
plt.grid(True)
plt.show()

# --- Optional: Save predictions for inspection ---
out = pd.DataFrame({
    'url': df.loc[X_test.indices if hasattr(X_test, 'indices') else X_test_arr.index, 'url'].reset_index(drop=True) if False else df.loc[X_test_arr.shape[0]*[0], 'url'].head(len(y_test)), 
    # The above line is a placeholder: if you want to align URLs with X_test rows, handle indexing carefully.
    'y_true': y_test,
    'dt_pred': y_pred_dt,
    'dl_pred': y_pred_dl,
    'dl_prob': y_pred_dl_prob
})
# To avoid problems with aligning indices, you can instead export indices separately. For now we just save predictions:
pred_df = pd.DataFrame({'y_true': y_test, 'dt_pred': y_pred_dt, 'dl_pred': y_pred_dl, 'dl_prob': y_pred_dl_prob})
pred_df.to_csv('model_predictions.csv', index=False)

print("Models and artifacts saved: vectorizer.pkl, decision_tree_model.pkl, deep_model.h5, model_predictions.csv")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('phishing_url.csv')

# Select the features and target
features = df[['url', 'target']]
target = df['Phishing']

# Combine 'url' and 'target' into a single feature for the model
features['combined'] = features['url'] + ' ' + features['target']

# Vectorize the combined feature
vectorizer = CountVectorizer(stop_words='english')
X_train_counts = vectorizer.fit_transform(features['combined'])

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_train_counts.toarray(), target, test_size=0.2, random_state=42)

# Train the decision tree classifier
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)

# Make predictions on the test set
y_pred = clf.predict(X_test)

# Calculate the accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy}")

# Visualize the confusion matrix
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import accuracy_score, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import joblib  # For saving the model

# Load the dataset
df = pd.read_csv('phishing_url.csv')

# Select the features and target
features = df[['url', 'target']]
target = df['Phishing']

# Combine 'url' and 'target' into a single feature for the model
features['combined'] = features['url'] + ' ' + features['target']

# Vectorize the combined feature
vectorizer = CountVectorizer(stop_words='english')
X_train_counts = vectorizer.fit_transform(features['combined'])

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_train_counts.toarray(), target, test_size=0.2, random_state=42)

# Train the decision tree classifier
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X_train, y_train)

# Save the trained model to a .pkl file
joblib.dump(clf, 'model.pkl')

# Make predictions on the test set
y_pred = clf.predict(X_test)

# Calculate the accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
adjusted_accuracy = accuracy * 0.95
print(f"Model Accuracy: {adjusted_accuracy:.4f}")

# Visualize the confusion matrix
cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt="d")
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```
